```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest

# =========================
# LOAD DATA
# =========================
df = pd.read_csv("/content/test.csv")

print("Data Loaded:", df.shape)

# Save time column separately
time_col = df["time"]

# =========================
# SORT
# =========================
df = df.sort_values(by=["PRN", "RX_time"]).reset_index(drop=True)

# =========================
# FEATURE ENGINEERING
# =========================
def feature_engineering(df):
    df = df.copy()

    df["EC_LC_ratio"] = df["EC"] / (df["LC"] + 1e-6)
    df["PIP_PQP_ratio"] = df["PIP"] / (df["PQP"] + 1e-6)
    df["signal_power"] = np.sqrt(df["PIP"]**2 + df["PQP"]**2)

    df["time_diff"] = df["RX_time"] - df["TOW"]

    df["doppler_diff"] = df.groupby("PRN")["Carrier_Doppler_hz"].diff().abs()
    df["phase_diff"] = df.groupby("PRN")["Carrier_phase"].diff().abs()
    df["range_diff"] = df.groupby("PRN")["Pseudorange_m"].diff().abs()

    df["doppler_roll_std"] = (
        df.groupby("PRN")["Carrier_Doppler_hz"]
        .rolling(5)
        .std()
        .reset_index(level=0, drop=True)
    )

    df["cn0_roll_mean"] = (
        df.groupby("PRN")["CN0"]
        .rolling(5)
        .mean()
        .reset_index(level=0, drop=True)
    )

    return df

df = feature_engineering(df)

# =========================
# CLEANING
# =========================
df = df.replace([np.inf, -np.inf], np.nan)
df = df.fillna(df.median(numeric_only=True))

# =========================
# RULE-BASED DETECTION
# =========================
doppler_th = df["doppler_diff"].mean() + 3 * df["doppler_diff"].std()
phase_th = df["phase_diff"].mean() + 3 * df["phase_diff"].std()
cn0_th = df["CN0"].mean() + 3 * df["CN0"].std()

df["rule_flag"] = (
    (df["doppler_diff"] > doppler_th) |
    (df["phase_diff"] > phase_th) |
    (df["CN0"] > cn0_th)
).astype(int)

# =========================
# MODEL INPUT
# =========================
X = df.select_dtypes(include=[np.number])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# =========================
# ISOLATION FOREST
# =========================
model = IsolationForest(
    n_estimators=200,
    contamination=0.1,
    random_state=42,
    n_jobs=-1
)

df["anomaly_score"] = model.fit_predict(X_scaled)

# Convert anomaly score
df["anomaly_flag"] = df["anomaly_score"].apply(lambda x: 1 if x == -1 else 0)

# =========================
# FINAL LABEL
# =========================
df["spoofed"] = ((df["rule_flag"] == 1) | (df["anomaly_flag"] == 1)).astype(int)

# =========================
# CONFIDENCE SCORE
# =========================
# Use anomaly score proxy
scores = model.decision_function(X_scaled)

# Normalize to 0–1
df["confidence"] = (scores - scores.min()) / (scores.max() - scores.min())

# =========================
# CREATE SUBMISSION
# =========================
submission = pd.DataFrame({
    "time": time_col,
    "spoofed": df["spoofed"],
    "confidence": df["confidence"]
})

submission.to_csv("submission.csv", index=False)
# Save file (ensure correct path)
submission.to_csv("/content/submission.csv", index=False)

# Download file
from google.colab import files
files.download("/content/submission.csv")

print("✅ Submission saved!")
print(submission.head())
```
